```python
from dace_query.sun import Sun

from astropy.io import fits

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta

import re  # find pattern in string
from tqdm import tqdm  # progress bar
import logging  # to avoid printing too many warnings

import tools as t
```

```python
# define the path of the directory where dace data is saved
path_dace = '../data/dace_sun/'
direc = 'Jun2018/'  # sub directory where the specific data is
# load absorption bands
bands = np.genfromtxt('../data/bands.txt', dtype=None, encoding=None, names=True, delimiter=',')
```

```python
# list of dates and max number of files per day
start = datetime(2015, 8, 1)

dates = [
    (start + timedelta(days=i)).strftime('%Y-%m-%d') for i in range(10)
    ]
print(dates)

# query
filters = {'date_night': {'equal': dates}}

data = Sun.query_database(
    filters=filters,
    output_format='pandas'
)

files = data['file_rootpath'].tolist()

print(f'Total files selected: {len(files)}')

Sun.download_files(
    's1d',
    files,
    output_directory=path_dace+direc,
    output_filename='Aug2015.tar.gz'
)
```
